**1. Import Numpy**

```python
import numpy as np # np is short for numpy, numpy stands for Numerical Python
```

**2. Python List vs Numpy Array**

**Normal python list**

```python
marks = [70, 30, 40, 80, 90]
print(marks)
```

**Numpy Array**

```python
marks = np.array([70, 30, 40, 80, 90])
print(marks)
print(marks+5)  #Numpy Array make cakculations fast
```

**3. Creating Numpy Arrays**

One Dimensional:

```python
marks = np.array([70, 30, 40, 80, 90])
print(marks)
```

Two Dimensional:

```python
marks = np.array([[70, 30, 40, 80, 90],
                  [10, 20, 50,80,30],
                  [60, 50, 90, 20, 70]])
print(marks)
```

**4. Array Properties**

```python
marks = np.array([[70, 30, 40, 80, 90],
                  [10, 20, 50,80,30],
                  [60, 50, 90, 20, 70]])
print(marks.ndim)
print(marks.shape)
print(marks.dtype)
print(marks.itemsize)
print(marks.size)
```

**5. Indexing**

```python
marks = np.array([70, 30, 40, 80, 90])
print(marks[0])
print(marks[4])
print(marks[-1])
print(marks[-5])

```

**6. Indexing in 2D-Array**

```python
marks = np.array([[70, 30, 40, 80, 90],
                  [10, 20, 50,80,30],
                  [60, 50, 90, 20, 70]])
print(marks)
print(marks[0,2])
print(marks[1, -2])
print(marks[2, 3])
#print(marks[2, 2:4])
```

**7. Slicing**

```python
marks = np.array([70, 30, 40, 80, 90])
print(marks[1:4])
print(marks[2:])
print(marks[:3])
print(marks[-3:-1])
```

```python
marks = np.array([[70, 30, 40, 80, 90],
                  [10, 20, 50,80,30]])
print(marks[0, 1:4])
print(marks[1, 2:4])
print(marks[1, 2:5])
```

**8. Vectorized operations**

```python
marks = np.array([70, 30, 40, 80, 90])
print(marks+5)
print(marks-5)
print(marks*5)
print(marks/5)
```

**9. Operations between two arrays**

```python
std1_marks = np.array([70, 30, 40, 80, 90])
std2_marks = np.array([10, 20, 50, 80, 30])
print(std1_marks+std2_marks)
print(std1_marks-std2_marks)
print(std1_marks*std2_marks)
print(std1_marks/std2_marks)
```

**10. Mathematical Functions**
Absolute, Round, Power and Square root

```python
std1_marks = np.array([70, 30, 40, 80, 90])
std2_marks = np.array([-10, 20, -50, 80, -30])
std3_marks = np.array([10.3, 20.5, 50.8, 80.9, 30.6])
print(np.sqrt(std1_marks))
print(np.power(std1_marks, 2))
print(np.abs(std2_marks))
print(np.round(std3_marks))
```

**11. Aggregation Function**
Sum, Mean, Min, Max, Median, Standard Deviation

```python
marks = np.array([70, 30, 40, 80, 90])
print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.median(marks))
print(np.std(marks))
```

**12. Conditions in NumPy**

```python
std1 = np.array([30, 60, 90, 80, 50])
print(std1>=50)
```

**13. Filtering Data**

```python
std1 = np.array([30, 60, 90, 80, 50])
print(std1[std1>=50])
```

**14. Reshape()**

```python
std_marks = np.array([30, 60, 90, 80, 50, 70])
print(std_marks)
print("-----reshaped array-----")
new_std_marks = std_marks.reshape(2, 3)
print(new_std_marks)
```

**15. Flatten()**

Changes Multi-dimensional array into one dimensional

```python
marks = np.array([[70, 30, 40, 80, 90],
                  [10, 20, 50,80,30],
                  [60, 50, 90, 20, 70]])  # number of value must be same in all the arrays
print(marks)
print("-----flatten array-----")
new_marks = marks.flatten()
print(new_marks)
```

**16. Useful NumPy Functions**

zeros(), ones(), arrange()

```python
# zeros()
print(np.zeros(10))

# ones()
print(np.ones(5))

# arange
print(np.arange(1,21))

# arange with step size
print(np.arange(2,21,2))
```

**18. Random Numbers**

```python
marks = np.random.randint(2, 50, 20) # it says generate 20 random numbers between 2 to 50. Numbers can be repeated in this.
print(marks)
```
